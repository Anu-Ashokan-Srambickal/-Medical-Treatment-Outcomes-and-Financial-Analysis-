### 1️⃣ - Yearly Billing Trends

```python
yb = df.groupby(df['Date of Admission'].dt.to_period('Y'))['Billing Amount'].sum()
yb
```

```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'], errors='coerce')
yearly_billing = df.groupby(df['Date of Admission'].dt.to_period('Y'))['Billing Amount'].sum()
plt.figure(figsize=(30,8))
yearly_billing.plot(kind='line', figsize=(12, 5), marker='o', color='crimson')
plt.title('Yearly Total Billing Amount',fontsize=20)
plt.ylabel('Billing Amount')
plt.xlabel('Year')
plt.show()
```

## Inferences:
  The line plot illustrates the yearly total billing amount, showing a significant increase from 2019 to 2020, followed by a period of stability, and then a dramatic drop in 2024.

Initial Growth (2019-2020):
The total billing amount sharply increased from around 1.75e8 (175 million) in 2019 to approximately 2.75e8 (275 million) in 2020.

Plateau (2020-2023): 
From 2020 through 2023, the total billing amount remained remarkably stable, consistently hovering around 2.75e8.

Sharp Decline (2023-2024): A very significant drop occurred from 2023 to 2024, bringing the total billing amount down to approximately 1.0e8 (100 million), marking the lowest point in the observed period.

This trend suggests a substantial increase in billing activity in the early years, a period of sustained high billing, and then a drastic reduction in the most recent year.

The heatmap shows the correlation coefficients between "Age," "Billing Amount," "Room Number," and "Length of Stay.

The color intensity indicates the strength and direction of the correlation, with red indicating a positive correlation, blue indicating a negative 
correlation, and values close to zero indicating a weak or no correlation.

The diagonal values are 1, as a variable is perfectly correlated with itself.

Conclusion:

The correlation matrix reveals that there are very weak to negligible correlations among "Age," "Billing Amount," "Room Number," and "Length of Stay.
    
All the off-diagonal values (correlations between different variables) are extremely close to zero, ranging from -0.0049 to 0.0079.

Specifically:

Age has almost no correlation with "Billing Amount" (-0.0034), "Room Number" (-0.0035), or "Length of Stay" (0.0079).

Billing Amount similarly shows virtually no correlation with "Room Number" (-0.0039) or "Length of Stay" (-0.0049).

Room Number has an extremely weak negative correlation with "Length of Stay" (-0.0045).

In essence, these variables appear to be largely independent of each other within this dataset. This suggests that changes in one variable, such as 
a person's age, do not predict or influence changes in another, like their billing amount or length of stay.
