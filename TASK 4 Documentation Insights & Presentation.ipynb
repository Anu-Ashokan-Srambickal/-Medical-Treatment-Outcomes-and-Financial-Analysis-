## The data preprocessing and feature engineering steps 

  1. Handing missing values (no missing values found).
     
  2. Removing duplicate rows.

  3. Formatted the 'Name' and 'Hospital' columns for consistency.

  4. Standardized 'Insurance Provider' names and converted 'Date of Admission' and 'Discharge Date' to datetime objects, and rounded 'Billing Amount'    to two decimal places.

  5. Created a new 'Length of Stay' column by calculating the difference between 'Discharge Date' and 'Date of Admission'.

  6. Filtered the data to show records where 'Length of Stay' is 25 days or more.

  7. Aggregated data to find the count of each gender, the average of numerical columns by 'Medical Condition', and the sum of numerical columns by  'Medication'.

  8. Saved the cleaned data to 'cleaned_HealthCareData.csv'.

## Methods

The following steps were undertaken for data cleaning and feature engineering:

🟠 Handling Missing Values: Checked for and confirmed the absence of missing values across all columns.

🟠 Removing Duplicates: Identified and removed duplicate rows to prevent skewed analysis and ensure each record represents unique information.

🟠 Formatting Data:

Stripped leading/trailing whitespace and converted 'Name' column entries to title case for uniformity.

    Cleaned the 'Hospital' column by stripping whitespace and applying specific string replacements to standardize hospital names 
    (e.g., re-arranging "and Montes Graves, Thomas" and removing "and" from the beginning/end).

🟠 Correcting Data Types:

Standardized 'Insurance Provider' names (e.g., 'UnitedHealthcare' to 'United Healthcare').

🟠 Converted 'Date of Admission' and 'Discharge Date' columns from string objects to datetime objects, enabling date-based calculations.

🟠 Rounded 'Billing Amount' to two decimal places for financial accuracy.

🟠 Creating Derived Columns:

Calculated 'Length of Stay' in days by subtracting 'Date of Admission' from 'Discharge Date'. This new feature is critical for understanding patient duration in hospitals.
